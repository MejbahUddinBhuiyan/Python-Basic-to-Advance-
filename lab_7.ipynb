```python
def up_lo(x):
  c1=0
  c2=0
  for i in x:
    if ord('A')<=ord(i)<=ord('Z'):
      c1=c1+1
    elif ord('a')<=ord(i)<=ord('z'):
      c2=c2+1
  print(f"No. of Uppercase characters : {c1}")
  print(f"No. of Lowercase Characters: {c2}")
up_lo(input())
```

```python
def concat(st,num):
  first=''
  last=''
  for i in range(len(st)):
    if i==0:
      first+=st[i]
    elif i%num==0:
      last+=st[i]
    else:
      first+=st[i]
  con=first+last
  return con
a=concat("I love programming.", 3)
print(a)
```
